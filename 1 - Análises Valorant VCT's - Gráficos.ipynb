```python
def plot_jogos_por_mapa(df, liga=None):
    if liga:
        df_filtrado = df[df['Liga'] == liga]
        titulo = f'Jogos por Mapa — {liga}'
    else:
        df_filtrado = df
        titulo = 'Jogos por Mapa — Todas as Ligas'

    mapas = df_filtrado['Mapa'].value_counts()
    if 'AllMaps' in mapas.index:
        mapas = mapas.drop('AllMaps')

    fig, ax = plt.subplots(figsize=(10, 6))
    cores = ['red' if v == mapas.max() else 'blue' for v in mapas.values]
    bars = ax.barh(mapas.index[::-1], mapas.values[::-1], color=cores[::-1], edgecolor='none', height=0.6)

    for bar, val in zip(bars, mapas.values[::-1]):
        ax.text(bar.get_width() + 0.2, bar.get_y() + bar.get_height() / 2,
                str(val), va='center', fontsize=12, fontweight='bold', color='white')

    ax.set_facecolor('#1a1a2e')
    fig.patch.set_facecolor('#1a1a2e')
    ax.tick_params(colors='white', labelsize=11)
    ax.spines[:].set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_title(titulo, color='white', fontsize=14, fontweight='bold', pad=15)
    plt.tight_layout()
    plt.show()


# Todas as ligas
plot_jogos_por_mapa(df)

# Por liga específica
plot_jogos_por_mapa(df, liga='VCT Americas')
plot_jogos_por_mapa(df, liga='VCT China')
plot_jogos_por_mapa(df, liga='VCT Emea')
plot_jogos_por_mapa(df, liga='VCT Pacifico')
```

```python
def plot_agentes(df, liga=None):
    if liga:
        df_filtrado = df[df['Liga'] == liga]
        titulo = f'Agentes Mais Escolhidos — {liga}'
    else:
        df_filtrado = df
        titulo = 'Agentes Mais Escolhidos — Todas as Ligas'

    agentes = df_filtrado['Agente'].value_counts()

    fig, ax = plt.subplots(figsize=(10, 6))
    cores = ['blue']
    bars = ax.barh(agentes.index[::-1], agentes.values[::-1], color=cores[::-1], edgecolor='none', height=0.6)

    for bar, val in zip(bars, agentes.values[::-1]):
        ax.text(bar.get_width() + 0.2, bar.get_y() + bar.get_height() / 2,
                str(val), va='center', fontsize=12, fontweight='bold', color='white')

    ax.set_facecolor('#1a1a2e')
    fig.patch.set_facecolor('#1a1a2e')
    ax.tick_params(colors='white', labelsize=11)
    ax.spines[:].set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_title(titulo, color='white', fontsize=14, fontweight='bold', pad=15)
    plt.tight_layout()
    plt.show()


# Todas as ligas
plot_agentes(df)

# Por liga
plot_agentes(df, liga='VCT Americas')
plot_agentes(df, liga='VCT China')
plot_agentes(df, liga='VCT Emea')
plot_agentes(df, liga='VCT Pacifico')
```

```python
def plot_top_jogadores(df, stat='R', top_n=5, liga=None):
    if liga:
        df_filtrado = df[df['Liga'] == liga]
        titulo = f'Jogadores por {stat} Médio — {liga}'
    else:
        df_filtrado = df
        titulo = f'Jogadores por {stat} Médio — Todas as Ligas'

    stat_por_jogador = (
        df_filtrado[df_filtrado['Mapa'] != 'AllMaps']
        .groupby('Jogador')[stat]
        .mean()
        .sort_values(ascending=False)
        .head(top_n)
    )

    n = len(stat_por_jogador)
    fig, ax = plt.subplots(figsize=(10, n * 0.5))
    cmap = plt.cm.get_cmap('RdYlGn', n + 2)
    cores = [cmap(i + 2) for i in range(n)]
    bars = ax.barh(stat_por_jogador.index[::-1], stat_por_jogador.values[::-1],
                   color=cores, edgecolor='none', height=0.4)

    for bar, val in zip(bars, stat_por_jogador.values[::-1]):
        ax.text(bar.get_width() + 0.02, bar.get_y() + bar.get_height() / 2,
                f'{val:.2f}', va='center', fontsize=11, fontweight='bold', color='white')

    ax.set_facecolor('#1a1a2e')
    fig.patch.set_facecolor('#1a1a2e')
    ax.tick_params(colors='white', labelsize=11)
    ax.spines[:].set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_title(titulo, color='white', fontsize=15, fontweight='bold', pad=5)
    plt.tight_layout()
    plt.show()


# Todas as ligas, top 5 por rating
plot_top_jogadores(df)

# Por liga
plot_top_jogadores(df, liga='VCT Americas')

# Outras stats
plot_top_jogadores(df, stat='ACS', top_n=10, liga='VCT Americas')
plot_top_jogadores(df, stat='K', top_n=10)
```

```python
def plot_rounds_por_time(df, liga=None):
    if liga:
        df_filtrado = df[df['Liga'] == liga]
        titulo = f'Rounds Totais por Time — {liga}'
    else:
        df_filtrado = df
        titulo = 'Rounds Totais por Time — Todas as Ligas'

    # 1 linha por partida (evita contar 10x o mesmo jogo)
    df_unico = df_filtrado.drop_duplicates(subset=['Time1', 'Time2', 'Mapa', 'Score_T1', 'Score_T2'])

    score_t1 = df_unico.groupby('Time1')['Score_T1'].sum()
    score_t2 = df_unico.groupby('Time2')['Score_T2'].sum()
    score_total = (score_t1
                   .add(score_t2, fill_value=0)
                   .sort_values(ascending=False)
                   .astype(int))

    if score_total.empty:
        print(f"Sem dados para: {titulo}")
        return

    fig, ax = plt.subplots(figsize=(10, len(score_total) * 0.5))
    cmap = plt.cm.get_cmap('RdYlGn', len(score_total) + 2)
    cores = [cmap(i + 2) for i in range(len(score_total))]
    bars = ax.barh(score_total.index[::-1], score_total.values[::-1],
                   color=cores, edgecolor='none', height=0.4)

    for bar, val in zip(bars, score_total.values[::-1]):
        ax.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height() / 2,
                str(val), va='center', fontsize=11, fontweight='bold', color='white')

    ax.set_facecolor('#1a1a2e')
    fig.patch.set_facecolor('#1a1a2e')
    ax.tick_params(colors='white', labelsize=11)
    ax.spines[:].set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_title(titulo, color='white', fontsize=15, fontweight='bold', pad=5)
    plt.subplots_adjust(top=0.97)
    plt.tight_layout()
    plt.show()


# Todas as ligas
plot_rounds_por_time(df)

# Por liga
plot_rounds_por_time(df, liga='VCT Americas')
plot_rounds_por_time(df, liga='VCT China')
plot_rounds_por_time(df, liga='VCT Emea')
plot_rounds_por_time(df, liga='VCT Pacifico')
```

```python
def plot_agente_por_jogador(df, liga=None):
    if liga:
        df_filtrado = df[df['Liga'] == liga]
        titulo = f'Agente Mais Jogado por Jogador — {liga}'
    else:
        df_filtrado = df
        titulo = 'Agente Mais Jogado por Jogador — Todas as Ligas'

    df_mapas = df_filtrado[df_filtrado['Mapa'] != 'AllMaps']

    agente_jogador = (df_mapas.groupby(['Jogador', 'Agente'])
                      .size()
                      .reset_index(name='Partidas')
                      .sort_values('Partidas', ascending=False))

    top_agente = (agente_jogador
                  .groupby('Jogador')
                  .first()
                  .reset_index()
                  .sort_values('Partidas', ascending=False))

    if top_agente.empty:
        print(f"Sem dados para: {titulo}")
        return

    fig, ax = plt.subplots(figsize=(12, len(top_agente) * 0.5))
    cmap = plt.cm.get_cmap('coolwarm', len(top_agente) + 2)
    cores = [cmap(i + 2) for i in range(len(top_agente))]
    bars = ax.barh(
        [f"{row.Jogador}  ({row.Agente})" for _, row in top_agente[::-1].iterrows()],
        top_agente['Partidas'].values[::-1],
        color=cores, edgecolor='none', height=0.4
    )

    for bar, val in zip(bars, top_agente['Partidas'].values[::-1]):
        ax.text(bar.get_width() + 0.1, bar.get_y() + bar.get_height() / 2,
                str(val), va='center', fontsize=10, fontweight='bold', color='white')

    ax.set_facecolor('#1a1a2e')
    fig.patch.set_facecolor('#1a1a2e')
    ax.tick_params(colors='white', labelsize=10)
    ax.spines[:].set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_title(titulo, color='white', fontsize=14, fontweight='bold', pad=5)
    plt.subplots_adjust(top=0.97)
    plt.tight_layout()
    plt.show()


# Todas as ligas
plot_agente_por_jogador(df)

# Por liga
plot_agente_por_jogador(df, liga='VCT Americas')
plot_agente_por_jogador(df, liga='VCT China')
```

```python
def plot_stats_agente(df, stat='Rating', min_partidas=5, liga=None):
    stat_label = {
        'Rating': 'Rating Médio',
        'ACS':    'ACS Médio',
        'KD':     'K/D Médio',
        'KAST':   'KAST Médio',
        'ADR':    'ADR Médio',
    }.get(stat, stat)

    if liga:
        df_filtrado = df[df['Liga'] == liga]
        titulo = f'{stat_label} por Agente — {liga}'
    else:
        df_filtrado = df
        titulo = f'{stat_label} por Agente — Todas as Ligas'

    df_mapas = df_filtrado[df_filtrado['Mapa'] != 'AllMaps']

    stats_agente = (df_mapas.groupby('Agente')
                    .agg(
                        Partidas = ('Agente', 'count'),
                        Rating   = ('R',      'mean'),
                        ACS      = ('ACS',    'mean'),
                        KD       = ('K',      'sum'),
                        Deaths   = ('D',      'sum'),
                        KAST     = ('KAST',   'mean'),
                        ADR      = ('ADR',    'mean'),
                    )
                    .assign(KD=lambda x: (x['KD'] / x['Deaths']).round(2))
                    .query(f'Partidas >= {min_partidas}')
                    .sort_values(stat, ascending=False)
                    .round(2))

    if stats_agente.empty:
        print(f"Sem dados para: {titulo}")
        return

    fig, ax = plt.subplots(figsize=(10, len(stats_agente) * 0.5))
    cmap = plt.cm.get_cmap('RdYlGn', len(stats_agente) + 2)
    cores = [cmap(i + 2) for i in range(len(stats_agente))]
    bars = ax.barh(stats_agente.index[::-1], stats_agente[stat].values[::-1],
                   color=cores, edgecolor='none', height=0.4)

    for bar, val in zip(bars, stats_agente[stat].values[::-1]):
        ax.text(bar.get_width() + 0.005, bar.get_y() + bar.get_height() / 2,
                f'{val:.2f}', va='center', fontsize=10, fontweight='bold', color='white')

    ax.set_facecolor('#1a1a2e')
    fig.patch.set_facecolor('#1a1a2e')
    ax.tick_params(colors='white', labelsize=10)
    ax.spines[:].set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_title(titulo, color='white', fontsize=14, fontweight='bold', pad=5)
    plt.tight_layout()
    plt.show()


# Todas as ligas, rating
plot_stats_agente(df)

# Por liga
plot_stats_agente(df, liga='VCT Americas')

# Outras stats
plot_stats_agente(df, stat='ACS', liga='VCT Americas')
plot_stats_agente(df, stat='KD', min_partidas=10)
plot_stats_agente(df, stat='ADR', liga='VCT China')
```

```python
def plot_melhor_por_mapa(df, stat='R', min_partidas=2, liga=None):
    stat_label = {
        'R':    'Rating',
        'ACS':  'ACS',
        'KD':   'K/D',
        'KAST': 'KAST',
        'ADR':  'ADR',
    }.get(stat, stat)

    if liga:
        df_filtrado = df[df['Liga'] == liga]
        titulo = f'Melhor Jogador por Mapa ({stat_label}) — {liga}'
    else:
        df_filtrado = df
        titulo = f'Melhor Jogador por Mapa ({stat_label}) — Todas as Ligas'

    df_mapas = df_filtrado[df_filtrado['Mapa'] != 'AllMaps']

    melhor_por_mapa = (df_mapas.groupby(['Mapa', 'Jogador'])
                       .agg(Stat=(stat, 'mean'), Partidas=(stat, 'count'))
                       .query(f'Partidas >= {min_partidas}')
                       .reset_index()
                       .sort_values('Stat', ascending=False)
                       .groupby('Mapa')
                       .first()
                       .reset_index()
                       .sort_values('Stat', ascending=False))

    if melhor_por_mapa.empty:
        print(f"Sem dados para: {titulo}")
        return

    fig, ax = plt.subplots(figsize=(10, len(melhor_por_mapa) * 0.6))
    cmap = plt.cm.get_cmap('coolwarm', len(melhor_por_mapa) + 2)
    cores = [cmap(i + 2) for i in range(len(melhor_por_mapa))]
    labels = [f"{row.Mapa}  →  {row.Jogador}" for _, row in melhor_por_mapa[::-1].iterrows()]

    bars = ax.barh(labels, melhor_por_mapa['Stat'].values[::-1],
                   color=cores, edgecolor='none', height=0.4)

    for bar, val in zip(bars, melhor_por_mapa['Stat'].values[::-1]):
        ax.text(bar.get_width() + 0.005, bar.get_y() + bar.get_height() / 2,
                f'{val:.2f}', va='center', fontsize=10, fontweight='bold', color='white')

    ax.set_facecolor('#1a1a2e')
    fig.patch.set_facecolor('#1a1a2e')
    ax.tick_params(colors='white', labelsize=10)
    ax.spines[:].set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_title(titulo, color='white', fontsize=14, fontweight='bold', pad=5)
    plt.tight_layout()
    plt.show()


# Todas as ligas, rating
plot_melhor_por_mapa(df)

# Por liga
plot_melhor_por_mapa(df, liga='VCT Americas')

# Outras stats
plot_melhor_por_mapa(df, stat='ACS', liga='VCT Americas')
plot_melhor_por_mapa(df, stat='ADR', min_partidas=3, liga='VCT China')
```

```python
def plot_stat_por_mapa(df, stat='K', liga=None):
    stat_label = {
        'K':    'Kills',
        'D':    'Deaths',
        'A':    'Assists',
        'R':    'Rating',
        'ACS':  'ACS',
        'ADR':  'ADR',
        'KAST': 'KAST',
    }.get(stat, stat)

    if liga:
        df_filtrado = df[df['Liga'] == liga]
        titulo = f'Média de {stat_label} por Mapa — {liga}'
    else:
        df_filtrado = df
        titulo = f'Média de {stat_label} por Mapa — Todas as Ligas'

    df_mapas = df_filtrado[df_filtrado['Mapa'] != 'AllMaps']

    stat_por_mapa = (df_mapas.groupby('Mapa')[stat]
                     .mean()
                     .sort_values(ascending=False)
                     .round(2))

    if stat_por_mapa.empty:
        print(f"Sem dados para: {titulo}")
        return

    fig, ax = plt.subplots(figsize=(10, len(stat_por_mapa) * 0.6))
    cmap = plt.cm.get_cmap('YlOrRd', len(stat_por_mapa) + 2)
    cores = [cmap(i + 2) for i in range(len(stat_por_mapa))]
    bars = ax.barh(stat_por_mapa.index[::-1], stat_por_mapa.values[::-1],
                   color=cores, edgecolor='none', height=0.4)

    for bar, val in zip(bars, stat_por_mapa.values[::-1]):
        ax.text(bar.get_width() + 0.05, bar.get_y() + bar.get_height() / 2,
                f'{val:.2f}', va='center', fontsize=11, fontweight='bold', color='white')

    ax.set_facecolor('#1a1a2e')
    fig.patch.set_facecolor('#1a1a2e')
    ax.tick_params(colors='white', labelsize=11)
    ax.spines[:].set_visible(False)
    ax.xaxis.set_visible(False)
    ax.set_title(titulo, color='white', fontsize=14, fontweight='bold', pad=5)
    plt.tight_layout()
    plt.show()


# Todas as ligas, kills
plot_stat_por_mapa(df)

# Por liga
plot_stat_por_mapa(df, liga='VCT Americas')

# Outras stats
plot_stat_por_mapa(df, stat='ACS', liga='VCT Americas')
plot_stat_por_mapa(df, stat='ADR', liga='VCT China')
plot_stat_por_mapa(df, stat='R')
```

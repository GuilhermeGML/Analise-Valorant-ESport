```python
df_geral = pd.read_csv('status_individual_geral.csv')
    
df_geral = df_geral.drop(
    columns=[
        'Time1',
        'Mapas_T1',
        'Score_T1',
        'Time2',
        'Mapas_T2',
        'Score_T2'
    ]
)
df_geral = df_geral.drop_duplicates()
df_geral = df_geral.reset_index(drop=True)

#Adicionar role(função) para cada jogador
agentes_valorant = {"duelistas": ["Jett", "Phoenix", "Reyna", "Raze", "Yoru", "Neon", "Iso", "Waylay"],
                    "controladores": ["Brimstone", "Viper", "Omen", "Astra", "Harbor", "Clove"],
                    "iniciadores": ["Sova", "Breach", "Skye", "Kayo", "Fade", "Gekko", "Tejo"],
                    "sentinelas": ["Sage", "Cypher", "Killjoy", "Chamber", "Deadlock", "Vyse", "Veto"]}

for i in range(len(df_geral)):
    agente_jogador = df_geral.loc[i, 'Agente']
    for role, agentes in agentes_valorant.items():
        if agente_jogador in agentes:
            df_geral.loc[i, 'Role'] = role

#Adicionar peso para cada função
peso_role = {
    'duelistas': {'R': 1.20, 'ACS': 1.20, 'K': 1.15, 'D': 1.00, 'A': 1.05, '+/-': 1.00, 'KAST': 1.10, 'ADR': 1.15, 'HS%': 1.00, 'FK': 1.15, 'FD': 1.00, '+/-_FK_FD': 1.00},
    'iniciadores': {'R': 1.15, 'ACS': 1.05, 'K': 1.10, 'D': 1.00, 'A': 1.20, '+/-': 1.00, 'KAST': 1.25, 'ADR': 1.20, 'HS%': 1.00, 'FK': 1.05, 'FD': 1.00, '+/-_FK_FD': 1.00},
    'controladores': {'R': 1.15, 'ACS': 1.05, 'K': 1.10, 'D': 1.00, 'A': 1.20, '+/-': 1.00, 'KAST': 1.25, 'ADR': 1.25, 'HS%': 1.00, 'FK': 1.00, 'FD': 1.00, '+/-_FK_FD': 1.00},
    'sentinelas': {'R': 1.15, 'ACS': 1.05, 'K': 1.10, 'D': 1.00, 'A': 1.15, '+/-': 1.00, 'KAST': 1.30, 'ADR': 1.20, 'HS%': 1.00, 'FK': 1.05, 'FD': 1.00, '+/-_FK_FD': 1.00}
}

colunas_stats = ['R', 'ACS', 'K', 'D', 'A', '+/-', 'KAST', 'ADR', 'HS%', 'FK', 'FD', '+/-_FK_FD']

for i in range(len(df_geral)):
    role_jogador = df_geral.iloc[i]['Role']
    pesos = peso_role[role_jogador]
    for coluna in colunas_stats:
        df_geral.loc[i, coluna] = df_geral.loc[i, coluna] * pesos[coluna]

df_geral[colunas_stats] = (df_geral[colunas_stats].round(3))

df_geral
```

```python
#Calculo de Impacto no time
# Tendo a certeza que a cada 5 linhas é um time eu posso calcular o impacto de cada jogador do time apenas com os dadosd daquela jodada,
# assim eu consigo ter mais noção de alguem q impacta sempre bastante, ou o contrário, assim é possivel ter 1 metrica para prever a % de win
# de um jogador

time = df_geral.iloc[70:75]

time['Impacto'] = (
    (time['R'] * 0.25) +
    (time['ACS'] * 0.20) +
    (time['K'] * 0.15) +
    (time['ADR'] * 0.15) +
    (time['KAST'] * 0.10) +
    (time['A'] * 0.10) +
    (time['FK'] * 0.05)
)
time['Impacto_Time_%'] = (
    time.groupby(['Mapa_Num', 'Time'])['Impacto']
    .transform(lambda x: (x / x.sum()) * 100)
)



time
```

```python
def time(j1, a1, j2, j3, j4, j5, mapa):

    t2 = pd.concat([

        df_impc[
            (df_impc['Mapa'] == mapa) &
            (df_impc['Jogador'] == j1)
        ],

        df_impc[
            (df_impc['Mapa'] == mapa) &
            (df_impc['Jogador'] == j2)&
            (df_impc['Agente'] == a1)
        ],

        df_impc[
            (df_impc['Mapa'] == mapa) &
            (df_impc['Jogador'] == j3)
        ],

        df_impc[
            (df_impc['Mapa'] == mapa) &
            (df_impc['Jogador'] == j4)
        ],

        df_impc[
            (df_impc['Mapa'] == mapa) &
            (df_impc['Jogador'] == j5)
        ]

    ], ignore_index=True)

    return t2

resultado = time(
    'aspas',
    'Kayo',
    'cauanzin',
    'lukxo',
    'Virtyy',
    'erde',
    'Icebox'
)

resultado    
```

```python
def gerar_time_aleatorio(df, n=5, seed=None):
    time = df.sample(n=n, random_state=seed).reset_index(drop=True)

    cols_soma = [
        'K', 'D', 'A', '+/-',
        'FK', 'FD', '+/-_FK_FD'
    ]

    cols_media = [
        'R', 'ACS', 'KAST',
        'ADR', 'HS%', 'Impacto'
    ]

    soma = time[cols_soma].sum()
    media = time[cols_media].mean()

    total = pd.concat([soma, media])

    total['Jogador'] = 'TOTAL'
    total['Mapa'] = '-'
    total['Time'] = '-'
    total['Agente'] = '-'
    total['Score_T1'] = '-'
    total['Score_T2'] = '-'
    total['target'] = '-'

    time = pd.concat([time, total.to_frame().T], ignore_index=True)

    return time


time = gerar_time_aleatorio(df_impc)
time
```

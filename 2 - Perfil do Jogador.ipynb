# Perfil do Jogadores

```python
def perfil_historico(nome, df, liga=None):
    if liga:
        df_filtrado = df[df['Liga'] == liga]
    else:
        df_filtrado = df

    jogador = df_filtrado[df_filtrado['Jogador'] == nome].copy()

    if jogador.empty:
        print(f"Jogador '{nome}' não encontrado" + (f" em {liga}" if liga else "") + "!")
        return

    cols_soma  = ['K', 'D', 'A', '+/-', 'FK', 'FD', '+/-_FK_FD']
    cols_media = ['R', 'ACS', 'KAST', 'ADR', 'HS%']

    soma  = jogador[cols_soma].sum()
    media = jogador[cols_media].mean()

    por_mapa = jogador.groupby('Mapa')[['ACS', 'target']].mean().round(3)
    por_mapa.columns = ['ACS_med', 'Win_Rate']
    por_mapa = por_mapa.sort_values('ACS_med', ascending=False)

    por_agente = jogador.groupby('Agente')[['ACS', 'target']].mean().round(3)
    por_agente.columns = ['ACS_med', 'Win_Rate']
    por_agente = por_agente.sort_values('ACS_med', ascending=False)

    liga_label = f" [{liga}]" if liga else ""
    print(f"=== Perfil Histórico: {nome}{liga_label} ===")
    print(f"Partidas : {len(jogador)}")
    print(f"Win Rate : {jogador['target'].mean():.1%}")
    print(f"Time(s)  : {', '.join(jogador['Time'].unique())}")
    print(f"Mapas    : {jogador['Mapa'].nunique()} diferentes")
    print()

    print("--- Médias ---")
    for col, val in media.items():
        print(f"  {col:<10}: {val:.3f}")
    print()

    print("--- Totais ---")
    for col, val in soma.items():
        print(f"  {col:<15}: {int(val)}")
    print()

    print("--- Por Mapa (ordenado por ACS) ---")
    for mapa, row in por_mapa.iterrows():
        print(f"  {mapa:<12} | ACS: {row['ACS_med']:.0f} | Win Rate: {row['Win_Rate']:.1%}")
    print()

    print("--- Por Agente (ordenado por ACS) ---")
    for agente, row in por_agente.iterrows():
        print(f"  {agente:<12} | ACS: {row['ACS_med']:.0f} | Win Rate: {row['Win_Rate']:.1%}")


# Uso
perfil_historico('cortezia', df)
#perfil_historico('cortezia', df, liga='VCT Americas')
```
